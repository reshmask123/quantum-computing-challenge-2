```python
# Qiskit Fall Fest 2026
# Challenge 2: Make the Quantum Coin Biased

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
import math

# Step 1: Create a quantum circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Step 2: Apply an Ry gate to create a biased quantum coin
theta = math.pi / 4
qc.ry(theta, 0)

# Step 3: Measure the qubit
qc.measure(0, 0)

# Step 4: Display the quantum circuit
print("Quantum Circuit:")
display(qc.draw(output='mpl'))

# Step 5: Predict measurement probabilities
prob_0 = math.cos(theta / 2) ** 2
prob_1 = math.sin(theta / 2) ** 2

print(f"Predicted probability of 0: {prob_0:.2%}")
print(f"Predicted probability of 1: {prob_1:.2%}")

# Step 6: Execute the circuit using 100 shots
simulator = AerSimulator()
job = simulator.run(qc, shots=100)
result = job.result()

# Step 7: Get measurement results
counts = result.get_counts()

print("\nMeasurement Results:")
print(counts)

# Step 8: Display the histogram
plot_histogram(counts)
plt.show()

# Step 9: Challenge question answer
print("\nChallenge Question:")
print(
    "Quantum gates change the state of a qubit and its measurement "
    "probabilities. The Ry gate rotates the qubit state by a specified "
    "angle, making the probabilities of measuring 0 and 1 unequal."
)
```
